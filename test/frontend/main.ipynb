```python
import requests
import pandas as pd

# Faz a requisição para o backend em Go
response = requests.get('http://localhost:8080/api/vendas')

# Transforma o JSON recebido em um DataFrame do Pandas
if response.status_code == 200:
    dados_json = response.json()
    df = pd.DataFrame(dados_json)
    print("Dados recebidos com sucesso!")
else:
    print(f"Erro ao conectar na API: {response.status_code}")
```

```python
import matplotlib.pyplot as plt

# Configura o gráfico usando os dados do DataFrame
plt.figure(figsize=(8, 4))
plt.bar(df['mes'], df['valor'], color='skyblue')

# Customização visual da interface
plt.title('Relatório de Vendas - Frontend Jupyter', fontsize=14)
plt.xlabel('Mês', fontsize=12)
plt.ylabel('Faturamento (R$)', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Exibe o resultado visual na tela do notebook
plt.show()
```
