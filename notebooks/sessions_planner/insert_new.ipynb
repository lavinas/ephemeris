```python
from sessions import create
import pandas as pd
def import_new(df):
    # Trata valores nulos (NaN) substituindo por string vazia
    df_clean = df.fillna("")
    
    for _, row in df_clean.iterrows():
        nickname = row["nickname"]
        if nickname == 'gabrielle':
            nickname = 'gabrielle_agostinho'
        result = create(
            nickname=str(row["nickname"]),
            date=str(row["date"]),
            minutes=int(row["minutes"]),
            service=str(row["service"]),
            status=str(row["status"]),
            comments=str(row["comments"])
        )
        print(result)
```
